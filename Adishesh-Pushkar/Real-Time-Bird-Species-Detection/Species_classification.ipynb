# Data Preprocessing & Augmentation (Bird Species Classification)

## Image Size
- Input size: **384 × 384**
- Chosen to retain fine-grained features while remaining computationally feasible
- Works well with EfficientNet-B4

---

## Training Augmentations

Applied **only during training** to reduce overfitting:

### Random Resized Crop
- Crops 70%–100% of the image area
- Encourages attention to different bird parts
- Avoids overly aggressive cropping

### Horizontal Flip
- Birds are left-right symmetric
- Improves robustness to pose variation

### Color Jitter
- Simulates changes in lighting conditions
- Small hue variation preserves species-specific colors

---

## Validation / Test Preprocessing

Validation and test pipelines are **fully deterministic**:

- Resize shorter side
- Center crop to final size
- No random augmentations

This ensures **stable and reproducible evaluation metrics**.

---

## Normalization

All images are normalized using **ImageNet statistics**:
- Required for pretrained EfficientNet weights
- Aligns feature distributions with pretraining data

Mean: `[0.485, 0.456, 0.406]`  
Std:  `[0.229, 0.224, 0.225]`

```python
from torchvision import transforms

# EfficientNet-B1 is designed for 240×240 inputs
IMG_SIZE = 240

train_tfms = transforms.Compose([
    # ---- Spatial robustness ----
    transforms.RandomResizedCrop(
        IMG_SIZE,
        scale=(0.7, 1.0),      # birds benefit from tighter crops than generic objects
        ratio=(0.75, 1.33)
    ),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomRotation(degrees=10),  # small pose variations

    # ---- Photometric robustness (kept light for birds) ----
    transforms.ColorJitter(
        brightness=0.1,
        contrast=0.1,
        saturation=0.1,
        hue=0.02
    ),

    transforms.ToTensor(),

    # ---- Normalization (ImageNet) ----
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    ),

    # ---- Occlusion robustness ----
    transforms.RandomErasing(
        p=0.25,
        scale=(0.02, 0.1),
        ratio=(0.3, 3.3),
        value="random"
    ),
])

val_tfms = transforms.Compose([
    transforms.Resize(IMG_SIZE + 32),   # 272
    transforms.CenterCrop(IMG_SIZE),    # 240
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])
```

```python
from torch.utils.data import Dataset

class BirdDataset(Dataset):
    def __init__(self, root, split, transform=None):
        self.img_dir = f"{root}/{split}/images"
        self.lbl_dir = f"{root}/{split}/labels"
        self.transform = transform
        self.files = sorted(os.listdir(self.img_dir))

    def __len__(self):
        return len(self.files)

    def __getitem__(self, idx):
        fname = self.files[idx]

        img = Image.open(f"{self.img_dir}/{fname}").convert("RGB")
        with open(f"{self.lbl_dir}/{fname.replace('.jpg','.txt')}") as f:
            label = int(f.read())

        if self.transform:
            img = self.transform(img)

        return img, label
```

```python
train_ds = BirdDataset(
    root="birdsnap_cls",
    split="train",
    transform=train_tfms
)

val_ds = BirdDataset(
    root="birdsnap_cls",
    split="val",
    transform=val_tfms
)
```

```python
from torch.utils.data import DataLoader

train_loader = DataLoader(
    train_ds,
    batch_size=16,
    shuffle=True,
    #num_workers=4,      # start with 2 or 4
    #pin_memory=True
)

val_loader = DataLoader(
    val_ds,
    batch_size=16,
    shuffle=False,
    #num_workers=4,
    #pin_memory=True
)
```

```python
import torch
import torch.nn as nn
from torchvision.models import efficientnet_b1, EfficientNet_B1_Weights

num_classes = 500
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Load ImageNet weights (for structure + missing buffers safety)
weights = EfficientNet_B1_Weights.IMAGENET1K_V1
model = efficientnet_b1(weights=weights)

# Rebuild classifier exactly as before
in_features = model.classifier[1].in_features

model.classifier = nn.Sequential(
    nn.Linear(in_features, 512),
    nn.BatchNorm1d(512),
    nn.ReLU(inplace=True),
    nn.Dropout(0.2),
    nn.Linear(512, num_classes)
)
```

```python
state_dict = torch.load(
    "species_classification-2.0.pth",
    map_location=device
)

model.load_state_dict(state_dict)
model.to(device)
model.eval()
```

```python
import torch.nn.functional as F

model.eval()

all_feats = []
all_probs = []
all_labels = []

with torch.no_grad():
    for images, labels in val_loader:
        images = images.to(device)
        labels = labels.to(device)

        # ---- Forward backbone ----
        feats = model.features(images)
        feats = F.adaptive_avg_pool2d(feats, 1).flatten(1)  # (B, D)

        outputs = model.classifier(feats)
        probs = torch.softmax(outputs, dim=1)

        all_feats.append(feats.cpu())
        all_probs.append(probs.cpu())
        all_labels.append(labels.cpu())

all_feats = torch.cat(all_feats)      # (N, D)
all_probs = torch.cat(all_probs)      # (N, C)
all_labels = torch.cat(all_labels)    # (N,)
```

```python
num_classes = all_probs.shape[1]
feat_dim = all_feats.shape[1]

prototypes = torch.zeros(num_classes, feat_dim)
counts = torch.zeros(num_classes)

for feat, label in zip(all_feats, all_labels):
    prototypes[label] += feat
    counts[label] += 1

prototypes /= counts.unsqueeze(1)
prototypes = F.normalize(prototypes, dim=1)  # cosine-ready
```

```python
torch.save(prototypes, "class_prototypes.pt")
```

```python
def feature_rerank_topk(feats, probs, prototypes, labels, k):
    feats = F.normalize(feats, dim=1)

    topk_idx = probs.topk(k, dim=1).indices  # (N, k)

    correct_top1 = 0
    correct_top5 = 0

    for i in range(len(labels)):
        feat = feats[i]
        cand_idx = topk_idx[i]          # candidate classes
        cand_proto = prototypes[cand_idx]

        sims = torch.matmul(cand_proto, feat)  # cosine similarity
        reranked = cand_idx[sims.argsort(descending=True)]

        if labels[i] == reranked[0]:
            correct_top1 += 1
        if labels[i] in reranked[:5]:
            correct_top5 += 1

    n = len(labels)
    return correct_top1 / n, correct_top5 / n
```

```python
rerank_top1, rerank_top5 = feature_rerank_topk(
    all_feats,
    all_probs,
    prototypes,
    all_labels,
    k=5
)

# print(f"Baseline Top-1 : {top1:.4f}")
# print(f"Baseline Top-5 : {top5:.4f}")
print("---- After Feature Reranking ----")
print(f"Reranked Top-1 : {rerank_top1:.4f}")
print(f"Reranked Top-5 : {rerank_top5:.4f}")
```
